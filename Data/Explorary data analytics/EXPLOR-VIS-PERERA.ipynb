```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
full_path = r"Perera_database - AQME DESCRIPTORS.xlsx"
df = pd.read_excel(full_path)
```

##### Data analytics

```python
# Yield distribution
sns.set_theme(style="ticks")
sns.color_palette()
f, ax = plt.subplots(figsize=(8, 5))
sns.histplot(df["yield_pct"], bins=30, edgecolor="0.2", linewidth=0.8, palette="deep", kde=True)
ax.axvline(df["yield_pct"].median(), color="red", linestyle="--", linewidth=1.2, label=f"Median={df['yield_pct'].median():.1f}%")
# ax.set_title("Yield Distribution", fontsize=14)
ax.set_xlabel("Yield (%)", fontsize=14)
ax.set_ylabel("Count", fontsize=14)
ax.tick_params(axis='both', labelsize=14)
legend = ax.legend(
    facecolor='white',      # Solid white background
    edgecolor='black',      # Black border
    framealpha=1,           # Fully opaque
    fancybox=True,          # Rounding of the box
)
legend.get_frame().set_linewidth(0.4) # Changing the thickness of the border
plt.savefig('Perera-Yield-Count.pdf', dpi=600, bbox_inches='tight')
```

```python
df["catalyst"].unique()
```

```python
# Yield by catalyst
median_order = df.groupby("catalyst")["yield_pct"].median().sort_values().index.tolist()
f, ax = plt.subplots(figsize=(13, 5))
order = sorted(df["catalyst"].unique())
sns.boxplot(data=df, x="catalyst", y="yield_pct", 
            linewidth=0.7, palette="deep", 
            saturation=1, linecolor="black", order=median_order)
sns.stripplot(data=df, x="catalyst", y="yield_pct", color="black", alpha=0.5, size=2.5)
ax.set_ylabel("Yield (%)", fontsize=14)
ax.set_xlabel("Ligand", fontsize=14)
ax.tick_params(axis='both', labelsize=14)
ax.grid(True, axis="y", linewidth=0.6, color="black", alpha=0.4)
plt.savefig('Perera-Yield-Catalyst.pdf', dpi=600,  bbox_inches='tight')
```

```python
# Yield by base
df['name_base'] = df['name_base'].fillna('No base')
median_order = df.groupby("name_base")["yield_pct"].median().sort_values().index.tolist()
f, ax = plt.subplots(figsize=(8, 5))
sns.boxplot(data=df, x="name_base", y="yield_pct", linewidth=0.7, palette="deep", 
            saturation=1, linecolor="black", order=median_order)
sns.stripplot(data=df, x="name_base", y="yield_pct", color="black", alpha=0.5, size=2.5)
# ax.set_title("Yield distribution by base", fontsize=14)
ax.set_ylabel("Yield (%)", fontsize=14)
ax.set_xlabel("Base", fontsize=14)
ax.tick_params(axis='both', labelsize=14)
ax.grid(True, axis="y", linewidth=0.6, color="black", alpha=0.4)
plt.savefig('Perera-Yield-Base.pdf', dpi=600,  bbox_inches='tight')
```

```python
# Yield by solvent
df['name_solvent'] = df['name_solvent'].fillna('No Solvent')
median_order = df.groupby("name_solvent")["yield_pct"].median().sort_values().index.tolist()
f, ax = plt.subplots(figsize=(8, 5))
sns.boxplot(data=df, x="name_solvent", y="yield_pct", linewidth=0.7, palette="deep", 
            saturation=1, linecolor="black", order=median_order)
sns.stripplot(data=df, x="name_solvent", y="yield_pct", color="black", alpha=0.5, size=2.5)
# ax.set_title("Yield distribution by base", fontsize=14)
ax.set_ylabel("Yield (%)", fontsize=14)
ax.set_xlabel("Solvent", fontsize=14)
ax.tick_params(axis='both', labelsize=14)
ax.grid(True, axis="y", linewidth=0.6, color="black", alpha=0.4)
plt.savefig('Perera-Yield-Solvent.pdf', dpi=600,  bbox_inches='tight')
```

```python
# Yield by boron species
f, ax = plt.subplots(figsize=(15, 5))
sns.boxplot(data=df, x="name_boron_species", y="yield_pct", 
            linewidth=0.7, palette="deep", 
            saturation=1, linecolor="black")
sns.stripplot(data=df, x="name_boron_species", y="yield_pct", color="black", alpha=0.4, size=2.5)
ax.set_title("Yield distribution by Boron Species", fontsize=14)
ax.set_ylabel("Yield (%)")
ax.set_xlabel("Boron Species")
```
